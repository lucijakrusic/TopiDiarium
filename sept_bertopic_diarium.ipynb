```python
import pandas as pd
from gensim.models.coherencemodel import CoherenceModel
from gensim.corpora import Dictionary

# --- Reuse your variables ---
# texts            -> list[str]
# timestamps       -> list[datetime-like or str]
# processed_docs   -> list[list[str]] from your preprocess()
# loaded_model     -> your BERTopic model

# 1) (Re)build a clean topics_list (strings only, no empty topics)
topic_words = loaded_model.get_topics()
topics_list = []
for tid, word_scores in topic_words.items():
    if tid == -1:
        continue
    words = [w for (w, _) in word_scores if isinstance(w, str) and len(w)]
    if words:
        topics_list.append(words)

# 2) Build a frame to group documents into the same monthly bins as your DTM labels (e.g., 'YYYY-MM')
df_coh = pd.DataFrame({
    "processed": processed_docs,
    "timestamp": pd.to_datetime(timestamps, errors="coerce")
})
df_coh = df_coh.dropna(subset=["timestamp"])
df_coh["bin"] = df_coh["timestamp"].dt.to_period("M").astype(str)

# 3) Compute c_v coherence per bin (filter topics to the bin’s vocabulary to avoid the Gensim error)
results = []
for bin_label, group in df_coh.groupby("bin", sort=True):
    bin_docs = group["processed"].tolist()

    # skip tiny bins
    if len(bin_docs) < 5:
        continue

    bin_dict = Dictionary(bin_docs)
    vocab = set(bin_dict.token2id.keys())

    # filter topic words to those present in this bin’s vocab; drop empty topics
    filtered_topics = [[w for w in t if w in vocab] for t in topics_list]
    filtered_topics = [t for t in filtered_topics if len(t) >= 2]

    # if no overlap, skip (prevents "unable to interpret topic..." error)
    if len(filtered_topics) < 2:
        continue

    cm = CoherenceModel(
        topics=filtered_topics,
        texts=bin_docs,
        dictionary=bin_dict,
        coherence="c_v"
    )
    score = cm.get_coherence()
    results.append({
        "time_bin": bin_label,
        "coherence_c_v": score,
        "n_docs": len(bin_docs),
        "vocab_size": len(vocab),
        "n_topics_used": len(filtered_topics)
    })


```

```python
# 4) Save results
coh_df = pd.DataFrame(results).sort_values("time_bin")
coh_df.to_csv("/content/drive/MyDrive/topic-specific-corpus-creation/csv_files/dtm_coherence.csv", index=False)
print(coh_df)
```

```python
average_coherence = coh_df["coherence_c_v"].mean()
print(f"\n📊 Average coherence over all bins: {average_coherence:.4f}")
```

```python
def calculate_topic_diversity(model, top_k=10):
    # Get topic words, skip topic -1 (outliers)
    topics = model.get_topics()

    topic_word_sets = [
        [word for word, _ in words[:top_k]]
        for topic_id, words in topics.items()
        if topic_id != -1 and words
    ]

    # Flatten the list of topic words
    all_words = [word for topic in topic_word_sets for word in topic]
    unique_words = set(all_words)

    # Compute diversity
    total_words = len(all_words)
    diversity = len(unique_words) / total_words if total_words > 0 else 0

    return diversity
```

```python
diversity_score = calculate_topic_diversity(loaded_model, top_k=10)
print(f"📚 Topic Diversity (top 10 words): {diversity_score:.4f}")
```

```python
filtered_doc_info = filtered_doc_info.merge(
    df_digi[["text", "date"]],
    left_on="Document",
    right_on="text",
    how="left"
)

```

```python
df_digi_sorted = df_digi.sort_values(by=["date"]).reset_index(drop=True)
df_digi_sorted["position"] = df_digi_sorted.index
text_to_index = df_digi_sorted.set_index("text")["position"].to_dict()
```

```python
def get_context(text, date):
    idx = text_to_index.get(text, None)
    if idx is None:
        return pd.Series([None, None])

    same_day_docs = df_digi_sorted[df_digi_sorted["date"] == date]
    same_day_texts = same_day_docs["text"].tolist()

    try:
        local_idx = same_day_texts.index(text)
    except ValueError:
        return pd.Series([None, None])

    # Get previous two and next two texts
    prev_docs = same_day_texts[max(0, local_idx - 2):local_idx]
    next_docs = same_day_texts[local_idx + 1:local_idx + 3]

    return pd.Series([
        " ".join(prev_docs),  # context_before
        " ".join(next_docs)   # context_after
    ])

# Step 6: Apply context function
filtered_doc_info[["context_before", "context_after"]] = filtered_doc_info.apply(
    lambda row: get_context(row["text"], row["date"]),
    axis=1
)
```

```python
final_df = filtered_doc_info[["text", "date", "Topic", "Representation", "context_before", "context_after"]]
```
